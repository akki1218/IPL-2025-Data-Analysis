```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
ipl=pd.read_excel("ipl_2025_Analysis.xlsx",skiprows=2)
ipl.head()
```

```python
ipl.info()
```

```python
#checking null values
ipl.isnull().sum()
```

```python
ipl[ipl.isnull().any(axis=1)]
```

```python
ipl = ipl.dropna()     #removing missing row
```

```python
ipl.shape
```

```python
ipl.isnull().sum()
```

```python

ipl = ipl[ipl['player of the match'] != 'Grand Total']
top_players = ipl.sort_values(by='Count of player_of_the_match',ascending=False).head(10)
top_players
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    top_players['player of the match'],
    top_players['Count of player_of_the_match']
)

plt.xlabel("Player")
plt.ylabel("Player of the Match Count")
plt.title("Top 10 Players of the Match - IPL 2025")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#batsman analysis
batsman=pd.read_excel("ipl_batsman.xlsx",skiprows=2)
batsman.head()
```

```python
batsman.columns
batsman.info()
```

```python
batsman = batsman[batsman['striker'] != 'Grand Total']
top_sixes = batsman.sort_values(by='Sum of sixes',ascending=False).head(10)
top_sixes
```

```python
plt.figure(figsize=(10, 6))
plt.bar(
    top_sixes['striker'],
    top_sixes['Sum of sixes']
)
plt.xlabel("Batsman")
plt.ylabel("Total Sixes")
plt.title("Top 10 Six Hitters - IPL 2025")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
#bowler analysis
bowler=pd.read_excel("ipl_bowler.xlsx",skiprows=2)
```

```python
bowler.head()
bowler.columns
```

```python
bowler.info()
```

```python
bowler = bowler[bowler['Bowler'] != 'Grand Total']
top_wicket_takers = bowler.sort_values(
    by='Sum of wickets_taken',
    ascending=False
).head(10)

top_wicket_takers
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    top_wicket_takers['Bowler'],
    top_wicket_takers['Sum of wickets_taken']
)

plt.xlabel("Bowler")
plt.ylabel("Total Wickets")
plt.title("Top 10 Wicket-Takers - IPL 2025")

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
